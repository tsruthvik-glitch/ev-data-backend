# OpenChargeMap Project

```python
data_pipeline/extract.py

import requests

def extract_data():
    response = requests.get("API_URL")
    return response.json()
```

```python
data_pipeline/transform.py

import pandas as pd

def transform_data(raw_data):
    df = pd.DataFrame(raw_data)
    df.dropna(inplace=True)
    return df
```

```python
data_pipeline/load.py

import sqlite3

def load_data(df):
    conn = sqlite3.connect("ev.db")
    df.to_sql("ev_data", conn, if_exists="replace")
```

##Build a Backend API (FastAPI Example)

```python
backend/main.py

from fastapi import FastAPI
from routes.ev_routes import router

app = FastAPI(title="EV Data Backend")
app.include_router(router)
```

```python
backend/routes/ev_routes.py

from fastapi import APIRouter
import sqlite3

router = APIRouter(prefix="/ev")

@router.get("/data")
def get_ev_data():
    conn = sqlite3.connect("ev.db")
    data = conn.execute("SELECT * FROM ev_data").fetchall()
    return {"records": data}
```
